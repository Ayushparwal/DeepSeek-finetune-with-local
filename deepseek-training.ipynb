```python
%%time
from ollama import chat

response = chat(model='deepseek-r1:1.5b', messages=[
  {
    'role': 'user',
    'content': 'Write 4 line on love',
  },
])
print(response.message.content)
```

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="deepseek-r1:1.5b",
    temperature=0.7,
)
```

```python
from langchain_core.messages import AIMessage

messages = [
    (
        "system",
        "You are a helpful assistant that can write creative poems",
    ),
    ("human", "give me the poem by robert frost roads like two road diverge in the yellow woods"),
]
ai_msg = llm.invoke(messages)
ai_msg
```

```python
message = ai_msg.content
```
